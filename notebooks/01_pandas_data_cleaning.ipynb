## 1. Load the dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Superstore_raw.csv", encoding="latin1")
print(f"Rows: {df.shape[0]:,} | Columns: {df.shape[1]}")
df.head()
```

## 3. Handle missing values

### 3a. Identify missing values

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct}).query("missing_count > 0")
```

Nearly every column has ~806 missing values concentrated at the **bottom of the file**. Let's look at those rows directly.

```python
df[df["Order Date"].isnull()][["Row ID", "Order ID", "Order Date"]].head(10)
```

**What I found:** rows from index 9994 onward aren't real orders at all — they're leftover rows from *other* sheets
in the original Excel workbook (a "People" sheet and a "Returns" sheet) that got concatenated into this CSV by
mistake. Their `Row ID` values are things like `"Person"`, `"Anna Andreadi"`, `"Returned"`, `"Yes"` — not numeric IDs.

Since these rows are structurally invalid for this table (not just "missing a value here and there"), I decided the
right fix was to **drop them entirely** rather than try to fill them in.

```python
# Row ID should always be numeric for genuine order rows — use that to isolate the real data
df["Row ID"] = pd.to_numeric(df["Row ID"], errors="coerce")
df = df[df["Row ID"].notnull()].copy()
df["Row ID"] = df["Row ID"].astype(int)
print(f"Shape after dropping non-order rows: {df.shape}")
```

```python
# Re-check missing values on the cleaned rows
df.isnull().sum()[df.isnull().sum() > 0]
```

### 3b. Fill the remaining missing values
Only `Postal Code` has real gaps now. Since postal code correlates with City/State, I filled each missing
value with the most common postal code for that City (falling back to 0 if a city has no other rows).

```python
city_postal = df.groupby("City")["Postal Code"].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
df["Postal Code"] = df.apply(
    lambda row: city_postal.get(row["City"]) if pd.isnull(row["Postal Code"]) else row["Postal Code"],
    axis=1
)
df["Postal Code"] = df["Postal Code"].fillna(0).astype(int)
print("Remaining missing values:", df.isnull().sum().sum())
```

### 3c. Fix data types
`Order Date` / `Ship Date` should be real dates, not strings.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%m/%d/%Y")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%m/%d/%Y")
df.dtypes[["Order Date", "Ship Date", "Postal Code"]]
```

## 4. Basic operations — filter rows & select columns

```python
# Filter: high-value Technology orders shipped Same Day
tech_same_day = df[(df["Category"] == "Technology") & (df["Ship Mode"] == "Same Day")]
print(f"Technology + Same Day orders: {len(tech_same_day)}")
tech_same_day[["Order ID", "Product Name", "Sales", "Quantity"]].head()
```

```python
# Select a subset of columns for a lightweight summary view
summary_cols = df[["Order ID", "Order Date", "Customer Name", "Category", "Sales", "Quantity", "Profit"]]
summary_cols.head()
```

## 5. Remove duplicates

```python
dupe_count = df.duplicated().sum()
print(f"Exact duplicate rows: {dupe_count}")
df = df.drop_duplicates()
print(f"Shape after de-duplication: {df.shape}")
```

## 6. Create a derived column — `total_amount`

The raw dataset stores `Sales` as the **line total** for each order row, not a per-unit price, so there's no
`price` column to multiply directly. To satisfy `total_amount = price * quantity` in a way that's honest about
the data, we first back out an implied **unit price**, then recompute `total_amount` from it (which should land
very close to the original `Sales` figure — a nice sanity check).

```python
df["unit_price"] = (df["Sales"] / df["Quantity"]).round(2)
df["total_amount"] = (df["unit_price"] * df["Quantity"]).round(2)

df[["Sales", "Quantity", "unit_price", "total_amount"]].head()
```

```python
# Sanity check: total_amount should reconcile with Sales (within rounding)
diff = (df["total_amount"] - df["Sales"]).abs()
print("Max discrepancy between total_amount and original Sales:", diff.max())
```

## 7. Save the cleaned dataset

```python
df.to_csv("Superstore_cleaned.csv", index=False)
print(f"Saved Superstore_cleaned.csv — {df.shape[0]:,} rows, {df.shape[1]} columns")
```
